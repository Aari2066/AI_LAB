```python
def board():
    return [' '] * 9

def show(b):
    print(f"\n{b[0]} | {b[1]} | {b[2]}")
    print("--+---+--")
    print(f"{b[3]} | {b[4]} | {b[5]}")
    print("--+---+--")
    print(f"{b[6]} | {b[7]} | {b[8]}\n")

def win(b, p):
    lines = [(0,1,2),(3,4,5),(6,7,8),
             (0,3,6),(1,4,7),(2,5,8),
             (0,4,8),(2,4,6)]
    return any(all(b[i] == p for i in line) for line in lines)

def empty(b):
    return [i for i in range(9) if b[i] == ' ']

def minimax(b, ai, human, maximizing):
    if win(b, ai): return 1
    if win(b, human): return -1
    if not empty(b): return 0

    scores = []
    for i in empty(b):
        b[i] = ai if maximizing else human
        scores.append(minimax(b, ai, human, not maximizing))
        b[i] = ' '

    return max(scores) if maximizing else min(scores)

def computer_move(b, ai, human):
    best, move = -2, None
    for i in empty(b):
        b[i] = ai
        score = minimax(b, ai, human, False)
        b[i] = ' '
        if score > best:
            best, move = score, i
    return move

def play():
    b = board()

    while True:
        human = input("Choose X or O: ").upper()
        if human in ['X', 'O']:
            break

    ai = 'O' if human == 'X' else 'X'
    turn = 'X'

    while True:
        show(b)

        if turn == human:
            while True:
                try:
                    move = int(input("Enter position (1-9): ")) - 1
                    if move in empty(b):
                        break
                    print("Invalid position.")
                except ValueError:
                    print("Enter a number.")
        else:
            move = computer_move(b, ai, human)
            print("Computer chose:", move + 1)

        b[move] = turn

        if win(b, turn):
            show(b)
            print(turn, "wins!")
            break

        if not empty(b):
            show(b)
            print("Draw!")
            break

        turn = 'O' if turn == 'X' else 'X'

play()
```

```python

room_A = input("Enter the state of Room A (Clean/Dirty): ").strip().capitalize()
room_B = input("Enter the state of Room B (Clean/Dirty): ").strip().capitalize()


position = input("Enter the starting room (A/B): ").strip().upper()

print("\n--- Vacuum Cleaner Iterations ---")

iteration = 1

while room_A == "Dirty" or room_B == "Dirty":

    print(f"\nIteration {iteration}")
    print(f"Vacuum is in Room {position}")
    print(f"Room A: {room_A}")
    print(f"Room B: {room_B}")

    if position == "A":
        if room_A == "Dirty":
            print("Action: Suck dirt from Room A")
            room_A = "Clean"
        else:
            print("Action: Room A is already clean")
            print("Action: Move to Room B")
            position = "B"

    elif position == "B":
        if room_B == "Dirty":
            print("Action: Suck dirt from Room B")
            room_B = "Clean"
        else:
            print("Action: Room B is already clean")
            print("Action: Move to Room A")
            position = "A"

    iteration += 1


print("\n--- Final Result ---")
print("Room A:", room_A)
print("Room B:", room_B)
print("Both rooms are clean!")
print(f"Total iterations: {iteration - 1}")
```
