```python

C1 = [(True, "Food", ("x",)),
      (False, "Likes", ("John", "x"))]

C2 = [(False, "Food", ("Apple",))]

C3 = [(False, "Food", ("Vegetables",))]

C4 = [(True, "Eats", ("x", "y")),
      (False, "Killed", ("x",)),
      (False, "Food", ("y",))]

C5 = [(False, "Eats", ("Anil", "Peanuts"))]

C6 = [(False, "Alive", ("Anil",))]

C7 = [(True, "Eats", ("Anil", "y")),
      (False, "Eats", ("Harry", "y"))]

C8 = [(True, "Alive", ("x",)),
      (True, "Killed", ("x",))]

C9 = [(False, "Killed", ("x",)),
      (False, "Alive", ("x",))]

C10 = [(True, "Likes", ("John", "Peanuts"))]


def substitute(term, subs):
    while term in subs:
        term = subs[term]
    return term


def unify(a, b, subs):
    a = substitute(a, subs)
    b = substitute(b, subs)

    if a == b:
        return True

    if a.islower():
        subs[a] = b
        return True

    if b.islower():
        subs[b] = a
        return True

    return False


def apply_substitution(literal, subs):
    neg, pred, args = literal
    args = tuple(substitute(a, subs) for a in args)
    return (neg, pred, args)


def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:
            neg1, pred1, args1 = l1
            neg2, pred2, args2 = l2

            if pred1 == pred2 and neg1 != neg2:
                subs = {}

                if len(args1) != len(args2):
                    continue

                valid = True

                for a, b in zip(args1, args2):
                    if not unify(a, b, subs):
                        valid = False
                        break

                if valid:
                    result = []

                    for lit in c1:
                        if lit != l1:
                            result.append(
                                apply_substitution(lit, subs)
                            )

                    for lit in c2:
                        if lit != l2:
                            result.append(
                                apply_substitution(lit, subs)
                            )

                    return result

    return None


def show(clause):
    if not clause:
        return "EMPTY CLAUSE"

    result = []

    for neg, pred, args in clause:
        sign = "~" if neg else ""
        result.append(
            sign + pred + "(" + ",".join(args) + ")"
        )

    return " OR ".join(result)


print("FIRST ORDER LOGIC - RESOLUTION")

print("\nNegated Query:")
print(show(C10))

R1 = resolve(C6, C8)
print("\nStep 1:", show(R1))

R2 = resolve(C4, C5)
print("\nStep 2:", show(R2))

R3 = resolve(R2, R1)
print("\nStep 3:", show(R3))

R4 = resolve(C1, R3)
print("\nStep 4:", show(R4))

R5 = resolve(R4, C10)
print("\nStep 5:", show(R5))

if R5 == []:
    print("\nQUERY PROVED")
    print("John likes Peanuts")
else:
    print("\nQUERY NOT PROVED")
```
