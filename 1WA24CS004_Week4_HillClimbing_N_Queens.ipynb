```python

def calculate_cost(board, n):
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def generate_neighbors(board, n):
    neighbors = []

    for col in range(n):
        for row in range(n):
            if board[col] != row:
                neighbor = board.copy()
                neighbor[col] = row
                neighbors.append(neighbor)

    return neighbors


def hill_climbing(board, n):
    current = board.copy()

    while True:
        current_cost = calculate_cost(current, n)

        print("\nCurrent State:", current)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("Goal State Reached!")
            return current

        neighbors = generate_neighbors(current, n)

        best_neighbor = min(neighbors, key=lambda x: calculate_cost(x, n))
        best_cost = calculate_cost(best_neighbor, n)

        print("Best Neighbor:", best_neighbor)
        print("Neighbor Cost:", best_cost)

        if best_cost >= current_cost:
            print("Local Minimum Reached!")
            return None

        current = best_neighbor


n = int(input("Enter number of queens: "))

initial = []

print("Enter row positions for each column (0 to", n - 1, "):")

for i in range(n):
    initial.append(int(input(f"Queen in column {i}: ")))

if any(row < 0 or row >= n for row in initial):
    print("Invalid row position!")
else:
    solution = hill_climbing(initial, n)

    if solution:
        print("\nFinal Solution:", solution)
        print("Final Cost:", calculate_cost(solution, n))
```
