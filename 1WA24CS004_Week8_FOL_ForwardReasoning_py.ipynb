```python

facts = [
    ("Eats", "Anil", "Peanuts"),
    ("Alive", "Anil"),
    ("Food", "Apple"),
    ("Food", "Vegetables")
]

rules = [
    ([("Alive", "?x")], ("NotKilled", "?x")),
    ([("Eats", "?x", "?y"),
      ("NotKilled", "?x")], ("Food", "?y")),
    ([("Food", "?x")], ("Likes", "John", "?x")),
    ([("Eats", "Anil", "?y")], ("Eats", "Harry", "?y"))
]


def match(pattern, fact, subs):
    if len(pattern) != len(fact):
        return None

    for p, f in zip(pattern, fact):
        if p.startswith("?"):
            if p in subs:
                if subs[p] != f:
                    return None
            else:
                subs[p] = f
        elif p != f:
            return None

    return subs


def apply(rule, subs):
    return tuple(
        subs.get(item, item) if item.startswith("?") else item
        for item in rule
    )


def forward_chaining(facts, rules):
    new_fact = True

    while new_fact:
        new_fact = False

        for premises, conclusion in rules:
            substitutions = [{}]

            for premise in premises:
                updated = []

                for sub in substitutions:
                    for fact in facts:
                        result = match(premise, fact, sub.copy())

                        if result is not None:
                            updated.append(result)

                substitutions = updated

            for sub in substitutions:
                result = apply(conclusion, sub)

                if result not in facts:
                    facts.append(result)
                    print("Derived:", result)
                    new_fact = True

    return facts


result = forward_chaining(facts, rules)

print("\nFinal Knowledge Base:")

for fact in result:
    print(fact)

query = ("Likes", "John", "Peanuts")

if query in result:
    print("\nQuery Proved: John likes Peanuts")
else:
    print("\nQuery Not Proved")
```
