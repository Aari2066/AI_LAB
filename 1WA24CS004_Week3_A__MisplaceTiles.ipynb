```python
import heapq

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)


def state_to_tuple(state):
    return tuple(tuple(row) for row in state)


def find_zero(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c


def get_neighbors(state):
    neighbors = []
    r, c = find_zero(state)

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in directions:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]

            new_state[r][c], new_state[nr][nc] = (
                new_state[nr][nc], new_state[r][c]
            )

            neighbors.append(state_to_tuple(new_state))

    return neighbors


def misplaced_tiles(state):
    count = 0

    for r in range(3):
        for c in range(3):
            if state[r][c] != 0 and state[r][c] != GOAL_STATE[r][c]:
                count += 1

    return count


def a_star(start_state):
    start_state = state_to_tuple(start_state)

    open_list = []
    heapq.heappush(
        open_list,
        (misplaced_tiles(start_state), 0, start_state)
    )

    came_from = {start_state: None}
    g_score = {start_state: 0}

    while open_list:
        f, g, current = heapq.heappop(open_list)

        if g != g_score.get(current):
            continue

        if current == GOAL_STATE:
            path = []
            state = current

            while state is not None:
                path.append(state)
                state = came_from[state]

            return path[::-1]

        for neighbor in get_neighbors(current):
            tentative_g = g + 1

            if tentative_g < g_score.get(neighbor, float("inf")):
                g_score[neighbor] = tentative_g
                came_from[neighbor] = current

                f_score = tentative_g + misplaced_tiles(neighbor)

                heapq.heappush(
                    open_list,
                    (f_score, tentative_g, neighbor)
                )

    return None


def print_board(state):
    for row in state:
        print(" ".join("_" if value == 0 else str(value) for value in row))
    print()


start_state = [
    [1, 2, 3],
    [0, 4, 6],
    [7, 5, 8]
]

print("--- A* Search using Misplaced Tiles Heuristic ---")
print("Initial State:")
print_board(start_state)

path = a_star(start_state)

if path:
    print("Solution found!")
    print("Number of moves:", len(path) - 1)
    print("Solution Path:")

    for step, state in enumerate(path):
        print("Step", step)
        print_board(state)
else:
    print("No solution found.")
```
