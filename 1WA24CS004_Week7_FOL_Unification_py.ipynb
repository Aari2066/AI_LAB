```python

def is_variable(x):
    return isinstance(x, str) and x.islower()


def occurs_check(var, value, subs):
    value = substitute(value, subs)

    if var == value:
        return True

    if isinstance(value, tuple):
        return any(occurs_check(var, item, subs) for item in value)

    return False


def substitute(x, subs):
    if is_variable(x) and x in subs:
        return substitute(subs[x], subs)

    if isinstance(x, tuple):
        return tuple(substitute(i, subs) for i in x)

    return x


def unify(x, y, subs):
    x = substitute(x, subs)
    y = substitute(y, subs)

    if x == y:
        return subs

    if is_variable(x):
        if occurs_check(x, y, subs):
            return None
        subs[x] = y
        return subs

    if is_variable(y):
        if occurs_check(y, x, subs):
            return None
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y):
            return None

        for a, b in zip(x, y):
            subs = unify(a, b, subs)

            if subs is None:
                return None

        return subs

    return None


x = ("Knows", "x", "John")
y = ("Knows", "Mary", "John")

result = unify(x, y, {})

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")
```
