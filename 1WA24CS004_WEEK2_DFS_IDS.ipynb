```python
import copy

GOAL_STATE = [[1, 2, 3],
              [4, 5, 6],
              [7, 8, 0]]


def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
    return -1, -1


def get_neighbors(state):
    neighbors = []
    r, c = find_zero(state)

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in directions:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            next_state = copy.deepcopy(state)
            next_state[r][c], next_state[nr][nc] = (
                next_state[nr][nc], next_state[r][c]
            )
            neighbors.append(next_state)

    return neighbors


def state_to_tuple(state):
    return tuple(tuple(row) for row in state)


def dfs(start_state, max_depth=100):
    stack = [(start_state, [start_state])]
    visited = {state_to_tuple(start_state)}

    while stack:
        current_state, path = stack.pop()
        depth = len(path) - 1

        if current_state == GOAL_STATE:
            return path

        if depth >= max_depth:
            continue

        for neighbor in get_neighbors(current_state):
            neighbor_tuple = state_to_tuple(neighbor)

            if neighbor_tuple not in visited:
                visited.add(neighbor_tuple)
                stack.append((neighbor, path + [neighbor]))

    return None


def depth_limited_search(state, depth, path, visited):
    if state == GOAL_STATE:
        return path

    if depth == 0:
        return None

    for neighbor in get_neighbors(state):
        neighbor_tuple = state_to_tuple(neighbor)

        if neighbor_tuple not in visited:
            visited.add(neighbor_tuple)

            result = depth_limited_search(
                neighbor, depth - 1, path + [neighbor], visited
            )

            if result is not None:
                return result

            visited.remove(neighbor_tuple)

    return None


def ids(start_state, max_depth=50):
    for depth in range(max_depth + 1):
        visited = {state_to_tuple(start_state)}

        result = depth_limited_search(
            start_state, depth, [start_state], visited
        )

        if result is not None:
            return result, depth

    return None, max_depth


def print_board(state):
    for row in state:
        print(" ".join(str(val) if val != 0 else "_" for val in row))
    print()


start_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

print("--- Initial State ---")
print_board(start_state)

print("--- 1. Solving with DFS ---")
dfs_path = dfs(start_state, max_depth=20)

if dfs_path:
    print(f"DFS found a solution in {len(dfs_path) - 1} moves.")
    print("Path:")

    for state in dfs_path:
        print_board(state)
else:
    print("DFS failed to find a solution within depth limit.")

print("--- 2. Solving with IDS ---")
ids_path, reached_depth = ids(start_state, max_depth=20)

if ids_path:
    print(
        f"IDS found an optimal solution at depth {reached_depth} "
        f"with {len(ids_path) - 1} moves."
    )
    print("Path:")

    for state in ids_path:
        print_board(state)
else:
    print("IDS failed to find a solution within max depth.")
```
